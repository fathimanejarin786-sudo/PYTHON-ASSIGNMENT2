```python
# 1. LIST CREATION

age_list = [24, 25, 26, 27, 28]

name_list = ["Anu", "Rahul", "Priya", "Arun", "Meera"]

print("Original age list:", age_list)
print("Original name list:", name_list)
```

```python
# 2. LIST OPERATIONS / MODIFICATIONS

name_list.append("Yazhini")
print("After appending:", name_list)

age_list.insert(2, 30)
print("After inserting 30:", age_list)

name_list.remove("Yazhini")
print("After removing Yazhini:", name_list)

age_list.pop()
print("After popping last element:", age_list)

age_list.extend([29, 30, 26])
print("After extending:", age_list)

age_list.sort(reverse=True)
print("Descending order:", age_list)

# g. Maximum, minimum and sum
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

```python
# 3. ACCESSING LIST ELEMENTS

print("First name:", name_list[0])
print("Last name:", name_list[-1])
print("Elements from index 2 to 4:", name_list[2:5])
print("Names in reverse:", name_list[::-1])
```

```python
# 4. DICTIONARY

student_marks = {
    "Anu": 85,
    "Rahul": 76,
    "Priya": 90,
    "Arun": 68,
    "Meera": 95
}

print("Student marks:", student_marks)

print("Priya's mark:", student_marks["Priya"])

student_marks["Janani"] = 80
print("After adding Janani:", student_marks)
student_marks["Rahul"] = 82
print("After updating Rahul:", student_marks)
print("Student names:", student_marks.keys())
print("Marks:", student_marks.values())
print("Student-mark pairs:", student_marks.items())
```

```python
# 5. SETS
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

print("My set:", my_set)

try:
    my_set[4] = 'z'
except TypeError as error:
    print("Error:", error)

set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```

```python
# 6. OPERATORS AND CONDITIONAL STATEMENTS

score = int(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
    print("Invalid score! Enter a value from 0 to 10.")

elif score > 7:
    print("Above Average: Excellent performance!")

elif score >= 4:
    print("Average: Good effort! Keep practicing, there's room for improvement.")

else:
    print("Below Average: Need to improve your performance. Consistent practice will lead to better results.")
```
